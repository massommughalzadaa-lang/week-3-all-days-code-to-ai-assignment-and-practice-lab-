### Exercise 3 — Is it even?
Write a function `is_even(n)` that returns `True` if `n` is even and `False` otherwise. Test it with 4 and 7.
*Hint: `return n % 2 == 0`.*

```python
# Your code here
def is_even(n):
    return n % 2 == 0

print(is_even(4))  # True
print(is_even(7))  # False
```

### Exercise 4 — Bigger of two
Write a function `bigger(a, b)` that returns the larger of the two numbers. Test with `bigger(10, 25)`.

```python
# Your code here
def bigger(a, b):
    if a > b:
        return a
    else:
        return b
print(bigger(20, 25))  # 25
```

### Exercise 5 — Sum a list
Write a function `total(numbers)` that takes a list and returns the sum of all its items (use a loop inside the function). Test with `[10, 20, 30]`.

```python
# Your code here
def total(numbers):
    result = 0
    for number in numbers:
        result += number
    return result
print(total([10, 20, 30]))  # 60
```

### Exercise 6 — Random dice 🎲
`import random`. Write a function `roll_dice()` that returns a random number between 1 and 6. Call it 5 times and print each roll.

```python
# Your code here
import random

def roll_dice():
    return random.randint(1, 6)

for i in range(5):
    print(roll_dice())
```

### Stretch challenge — Guess helper
Write a function `check_guess(guess, secret)` that returns the string `"low"`, `"high"`, or `"correct"` by comparing `guess` to `secret`. This is the brain of tomorrow's-style game — test it with a few values!

```python
# Your code here
def check_guess(guess, secret):
    if guess < secret:
        return "low!"
    elif guess > secret:
        return "high!"
    else:
        return "correct"
    
print(check_guess(5, 10))  # low!
print(check_guess(15, 10))  # high!
print(check_guess(10, 10))  # correct
print("test") 
    
```
