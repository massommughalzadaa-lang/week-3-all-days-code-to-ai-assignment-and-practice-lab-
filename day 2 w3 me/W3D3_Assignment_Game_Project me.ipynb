## Main Task — Number-Guessing Game

Build a working Number-Guessing Game inside a function `play_game()`:

1. The computer picks a random number between 1 and 100.
2. Keep asking the player to guess (use a loop).
3. After each guess, print `"Too high"` or `"Too low"`.
4. Count the number of guesses.
5. When correct, congratulate the player and show the guess count.

Start from what we built in class, then make it your own.

```python
import random

def play_game():
    print("Welcome to the Number Guessing Game!")
    secret = random.randint(1, 100)
    guesses = 0

    while True:
        guess = int(input("guess a number (between 1 and 100): "))
        guesses += 1

        if guess < secret:
            print("low")
        elif guess > secret:
            print("high")
        else:
            print(f"Congratulations!you got it in {guesses} guesses.")
            break

play_game()

```

Add at least one of these features to your game:

```python
import random

def play_game():
    secret = random.randint(1, 100)
    guesses = 0
    max_guesses = 7

    print("Welcome to the Number Guessing Game!")
    print(f"You have {max_guesses} tries to guess the number between 1 and 100.")

    while guesses < max_guesses:
        guess = int(input("guess a number (between 1 and 100): "))
        guesses += 1

        if guess < secret:
            print("low")
        elif guess > secret:
            print("high")
        else:
            print(f"Congratulations! You got it in {guesses} guesses.")
            break
    else:
        print(f"Game over! The number was {secret}.")

while True:
    play_game()
    play_again = input("play again? (yes/no): ")
    if play_again != 'yes':
        print("Thanks for playing!")
        break
```
